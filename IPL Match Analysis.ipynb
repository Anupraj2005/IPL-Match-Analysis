```python
import numpy as np
import pandas as pd

# ==========================================
# IPL MATCH ANALYSIS PROJECT (2022-2026)
# Using Python, Pandas and NumPy
# ==========================================


# -------------------- SEASON DATA --------------------

season_data = [

    ["2022", "Gujarat Titans", "Rajasthan Royals",
     "Jos Buttler (RR)", 863, "Yuzvendra Chahal (RR)", 27, 2017, 1062],

    ["2023", "Chennai Super Kings", "Gujarat Titans",
     "Shubman Gill (GT)", 890, "Mohammed Shami (GT)", 28, 2174, 1124],

    ["2024", "Kolkata Knight Riders", "Sunrisers Hyderabad",
     "Virat Kohli (RCB)", 741, "Harshal Patel (PBKS)", 24, 2174, 1260],

    ["2025", "Royal Challengers Bengaluru", "Punjab Kings",
     "Sai Sudharsan (GT)", 759, "Prasidh Krishna (GT)", 25, 2245, 1294],

    ["2026", "Royal Challengers Bengaluru", "Gujarat Titans",
     "Vaibhav Sooryavanshi (RR)", 776, "Kagiso Rabada (GT)", 29, 2332, 1426]
]

season_df = pd.DataFrame(season_data, columns=[
    "Year", "Winner", "Runner-up", "Orange Cap",
    "Orange Cap Runs", "Purple Cap", "Purple Cap Wickets",
    "Fours", "Sixes"
])

# Calculate total boundaries using Pandas
season_df["Total Boundaries"] = (
    season_df["Fours"] + season_df["Sixes"]
)


# -------------------- PLAYOFF DATA --------------------

playoff_data = [

    # IPL 2022 Playoffs
    ["2022", "Qualifier 1", "RR 188/6; GT 191/3",
     "Eden Gardens, Kolkata", "Gujarat Titans",
     "GT won by 7 wickets", "David Miller (GT)"],

    ["2022", "Eliminator", "RCB 207/4; LSG 193/6",
     "Eden Gardens, Kolkata", "Royal Challengers Bangalore",
     "RCB won by 14 runs", "Rajat Patidar (RCB)"],

    ["2022", "Qualifier 2", "RCB 157/8; RR 161/3",
     "Narendra Modi Stadium, Ahmedabad", "Rajasthan Royals",
     "RR won by 7 wickets", "Jos Buttler (RR)"],

    ["2022", "Final", "RR 130/9; GT 133/3",
     "Narendra Modi Stadium, Ahmedabad", "Gujarat Titans",
     "GT won by 7 wickets", "Hardik Pandya (GT)"],

    # IPL 2023 Playoffs
    ["2023", "Qualifier 1", "CSK 172/7; GT 157",
     "M. A. Chidambaram Stadium, Chennai", "Chennai Super Kings",
     "CSK won by 15 runs", "Ruturaj Gaikwad (CSK)"],

    ["2023", "Eliminator", "LSG 101; MI 182/8",
     "M. A. Chidambaram Stadium, Chennai", "Mumbai Indians",
     "MI won by 81 runs", "Akash Madhwal (MI)"],

    ["2023", "Qualifier 2", "GT 233/3; MI 171",
     "Narendra Modi Stadium, Ahmedabad", "Gujarat Titans",
     "GT won by 62 runs", "Shubman Gill (GT)"],

    ["2023", "Final", "GT 214/4; CSK 171/5",
     "Narendra Modi Stadium, Ahmedabad", "Chennai Super Kings",
     "CSK won by 5 wickets (DLS)", "Devon Conway (CSK)"],

    # IPL 2024 Playoffs
    ["2024", "Qualifier 1", "SRH 159; KKR 164/2",
     "Narendra Modi Stadium, Ahmedabad", "Kolkata Knight Riders",
     "KKR won by 8 wickets", "Mitchell Starc (KKR)"],

    ["2024", "Eliminator", "RCB 172/8; RR 174/6",
     "Narendra Modi Stadium, Ahmedabad", "Rajasthan Royals",
     "RR won by 4 wickets", "Ravichandran Ashwin (RR)"],

    ["2024", "Qualifier 2", "SRH 175/9; RR 139/7",
     "M. A. Chidambaram Stadium, Chennai", "Sunrisers Hyderabad",
     "SRH won by 36 runs", "Shahbaz Ahmed (SRH)"],

    ["2024", "Final", "SRH 113; KKR 114/2",
     "M. A. Chidambaram Stadium, Chennai", "Kolkata Knight Riders",
     "KKR won by 8 wickets", "Mitchell Starc (KKR)"],

    # IPL 2025 Playoffs
    ["2025", "Qualifier 1", "PBKS 101; RCB 106/2",
     "Maharaja Yadavindra Singh Stadium, Mullanpur",
     "Royal Challengers Bengaluru", "RCB won by 8 wickets",
     "Josh Hazlewood (RCB)"],

    ["2025", "Eliminator", "MI 228/5; GT 208/6",
     "Maharaja Yadavindra Singh Stadium, Mullanpur",
     "Mumbai Indians", "MI won by 20 runs", "Rohit Sharma (MI)"],

    ["2025", "Qualifier 2", "MI 203/6; PBKS 207/5",
     "Narendra Modi Stadium, Ahmedabad", "Punjab Kings",
     "PBKS won by 5 wickets", "Shreyas Iyer (PBKS)"],

    ["2025", "Final", "RCB 190/9; PBKS 184/7",
     "Narendra Modi Stadium, Ahmedabad",
     "Royal Challengers Bengaluru", "RCB won by 6 runs",
     "Krunal Pandya (RCB)"],

    # IPL 2026 Playoffs
    ["2026", "Qualifier 1", "RCB 254/5; GT 162",
     "HPCA Stadium, Dharamshala", "Royal Challengers Bengaluru",
     "RCB won by 92 runs", "Rajat Patidar (RCB)"],

    ["2026", "Eliminator", "RR 243/8; SRH 196",
     "Maharaja Yadavindra Singh Stadium, New Chandigarh",
     "Rajasthan Royals", "RR won by 47 runs",
     "Vaibhav Sooryavanshi (RR)"],

    ["2026", "Qualifier 2", "RR 214/6; GT 219/3",
     "Maharaja Yadavindra Singh Stadium, New Chandigarh",
     "Gujarat Titans", "GT won by 7 wickets", "Shubman Gill (GT)"],

    ["2026", "Final", "GT 155/8; RCB 161/5",
     "Narendra Modi Stadium, Ahmedabad",
     "Royal Challengers Bengaluru", "RCB won by 5 wickets",
     "Virat Kohli (RCB)"]
]

playoff_df = pd.DataFrame(playoff_data, columns=[
    "Year", "Match", "Scorecard", "Venue",
    "Winner", "Result", "Player of the Match"
])


# -------------------- PLAYER DATA --------------------

player_data = [

    # IPL 2022 - Run scorers
    ["2022", "Runs", 1, "Jos Buttler (RR)", 863],
    ["2022", "Runs", 2, "KL Rahul (LSG)", 616],
    ["2022", "Runs", 3, "Quinton de Kock (LSG)", 508],
    ["2022", "Runs", 4, "Hardik Pandya (GT)", 487],
    ["2022", "Runs", 5, "Shubman Gill (GT)", 483],

    # IPL 2022 - Wicket takers
    ["2022", "Wickets", 1, "Yuzvendra Chahal (RR)", 27],
    ["2022", "Wickets", 2, "Wanindu Hasaranga (RCB)", 26],
    ["2022", "Wickets", 3, "Kagiso Rabada (PBKS)", 23],
    ["2022", "Wickets", 4, "Umran Malik (SRH)", 22],
    ["2022", "Wickets", 5, "Kuldeep Yadav (DC)", 21],

    # IPL 2022 - Six hitters
    ["2022", "Sixes", 1, "Jos Buttler (RR)", 45],
    ["2022", "Sixes", 2, "Liam Livingstone (PBKS)", 34],
    ["2022", "Sixes", 3, "Andre Russell (KKR)", 32],
    ["2022", "Sixes", 4, "KL Rahul (LSG)", 30],
    ["2022", "Sixes", 5, "Sanju Samson (RR)", 26],

    # IPL 2023 - Run scorers
    ["2023", "Runs", 1, "Shubman Gill (GT)", 890],
    ["2023", "Runs", 2, "Faf du Plessis (RCB)", 730],
    ["2023", "Runs", 3, "Devon Conway (CSK)", 672],
    ["2023", "Runs", 4, "Virat Kohli (RCB)", 639],
    ["2023", "Runs", 5, "Yashasvi Jaiswal (RR)", 625],

    # IPL 2023 - Wicket takers
    ["2023", "Wickets", 1, "Mohammed Shami (GT)", 28],
    ["2023", "Wickets", 2, "Mohit Sharma (GT)", 27],
    ["2023", "Wickets", 3, "Rashid Khan (GT)", 27],
    ["2023", "Wickets", 4, "Piyush Chawla (MI)", 22],
    ["2023", "Wickets", 5, "Yuzvendra Chahal (RR)", 21],

    # IPL 2023 - Six hitters
    ["2023", "Sixes", 1, "Faf du Plessis (RCB)", 36],
    ["2023", "Sixes", 2, "Shivam Dube (CSK)", 35],
    ["2023", "Sixes", 3, "Shubman Gill (GT)", 33],
    ["2023", "Sixes", 4, "Glenn Maxwell (RCB)", 31],
    ["2023", "Sixes", 5, "Ruturaj Gaikwad (CSK)", 30],

    # IPL 2024 - Run scorers
    ["2024", "Runs", 1, "Virat Kohli (RCB)", 741],
    ["2024", "Runs", 2, "Ruturaj Gaikwad (CSK)", 583],
    ["2024", "Runs", 3, "Riyan Parag (RR)", 573],
    ["2024", "Runs", 4, "Travis Head (SRH)", 567],
    ["2024", "Runs", 5, "Sanju Samson (RR)", 531],

    # IPL 2024 - Wicket takers
    ["2024", "Wickets", 1, "Harshal Patel (PBKS)", 24],
    ["2024", "Wickets", 2, "Varun Chakaravarthy (KKR)", 21],
    ["2024", "Wickets", 3, "Jasprit Bumrah (MI)", 20],
    ["2024", "Wickets", 4, "T Natarajan (SRH)", 19],
    ["2024", "Wickets", 5, "Harshit Rana (KKR)", 19],

    # IPL 2024 - Six hitters
    ["2024", "Sixes", 1, "Abhishek Sharma (SRH)", 42],
    ["2024", "Sixes", 2, "Virat Kohli (RCB)", 38],
    ["2024", "Sixes", 3, "Heinrich Klaasen (SRH)", 38],
    ["2024", "Sixes", 4, "Nicholas Pooran (LSG)", 36],
    ["2024", "Sixes", 5, "Riyan Parag (RR)", 33],

    # IPL 2025 - Run scorers
    ["2025", "Runs", 1, "Sai Sudharsan (GT)", 759],
    ["2025", "Runs", 2, "Suryakumar Yadav (MI)", 717],
    ["2025", "Runs", 3, "Virat Kohli (RCB)", 657],
    ["2025", "Runs", 4, "Shubman Gill (GT)", 650],
    ["2025", "Runs", 5, "Mitchell Marsh (LSG)", 627],

    # IPL 2025 - Wicket takers
    ["2025", "Wickets", 1, "Prasidh Krishna (GT)", 25],
    ["2025", "Wickets", 2, "Noor Ahmad (CSK)", 24],
    ["2025", "Wickets", 3, "Josh Hazlewood (RCB)", 22],
    ["2025", "Wickets", 4, "Trent Boult (MI)", 22],
    ["2025", "Wickets", 5, "Arshdeep Singh (PBKS)", 21],

    # IPL 2025 - Six hitters
    ["2025", "Sixes", 1, "Nicholas Pooran (LSG)", 40],
    ["2025", "Sixes", 2, "Shreyas Iyer (PBKS)", 39],
    ["2025", "Sixes", 3, "Suryakumar Yadav (MI)", 38],
    ["2025", "Sixes", 4, "Mitchell Marsh (LSG)", 37],
    ["2025", "Sixes", 5, "Prabhsimran Singh (PBKS)", 30],

    # IPL 2026 - Run scorers
    ["2026", "Runs", 1, "Vaibhav Sooryavanshi (RR)", 776],
    ["2026", "Runs", 2, "Shubman Gill (GT)", 732],
    ["2026", "Runs", 3, "Sai Sudharsan (GT)", 722],
    ["2026", "Runs", 4, "Virat Kohli (RCB)", 675],
    ["2026", "Runs", 5, "Heinrich Klaasen (SRH)", 624],

    # IPL 2026 - Wicket takers
    ["2026", "Wickets", 1, "Kagiso Rabada (GT)", 29],
    ["2026", "Wickets", 2, "Bhuvneshwar Kumar (RCB)", 28],
    ["2026", "Wickets", 3, "Jofra Archer (RR)", 25],
    ["2026", "Wickets", 4, "Rashid Khan (GT)", 21],
    ["2026", "Wickets", 5, "Anshul Kamboj (CSK)", 21],

    # IPL 2026 - Six hitters
    ["2026", "Sixes", 1, "Vaibhav Sooryavanshi (RR)", 72],
    ["2026", "Sixes", 2, "Abhishek Sharma (SRH)", 43],
    ["2026", "Sixes", 3, "Rajat Patidar (RCB)", 42],
    ["2026", "Sixes", 4, "Ryan Rickelton (MI)", 38],
    ["2026", "Sixes", 5, "Mitchell Marsh (LSG)", 36]
]

player_df = pd.DataFrame(player_data, columns=[
    "Year", "Category", "Rank", "Player", "Value"
])


# -------------------- FUNCTIONS --------------------


# 1. Display season summary
def show_season_summary():

    year = input("Enter year (2022, 2023, 2024, 2025 or 2026): ")

    data = season_df[season_df["Year"] == year]

    if data.empty:
        print("Invalid year.")

    else:
        print("\nIPL", year, "SEASON SUMMARY")

        row = data.iloc[0]

        print("Winner:", row["Winner"])
        print("Runner-up:", row["Runner-up"])
        print("Orange Cap:", row["Orange Cap"])
        print("Orange Cap Runs:", row["Orange Cap Runs"])
        print("Purple Cap:", row["Purple Cap"])
        print("Purple Cap Wickets:", row["Purple Cap Wickets"])
        print("Total Fours:", row["Fours"])
        print("Total Sixes:", row["Sixes"])
        print("Total Boundaries:", row["Total Boundaries"])


# 2. Display playoff results
def show_playoff_results():

    year = input("Enter year (2022, 2023, 2024, 2025 or 2026): ")

    data = playoff_df[playoff_df["Year"] == year]

    print("\nPLAYOFF RESULTS OF", year)

    if data.empty:
        print("No data found.")

    else:
        for index, row in data.iterrows():

            print("\nMatch:", row["Match"])
            print("Scorecard:", row["Scorecard"])
            print("Venue:", row["Venue"])
            print("Winner:", row["Winner"])
            print("Result:", row["Result"])
            print("Player of the Match:", row["Player of the Match"])


# 3. Display top five players
def show_top_players():

    year = input("Enter year (2022, 2023, 2024, 2025 or 2026): ")

    print("\nSELECT CATEGORY")
    print("1. Top Run Scorers")
    print("2. Top Wicket Takers")
    print("3. Top Six Hitters")

    choice = input("Enter your choice: ")

    if choice == "1":
        category = "Runs"

    elif choice == "2":
        category = "Wickets"

    elif choice == "3":
        category = "Sixes"

    else:
        print("Invalid choice.")
        return

    data = player_df[
        (player_df["Year"] == year) &
        (player_df["Category"] == category)
    ]

    print("\nTOP 5", category, "OF", year)

    if data.empty:
        print("No data found.")

    else:
        for index, row in data.iterrows():

            print(row["Rank"], row["Player"], "-", row["Value"])


# 4. Compare boundaries using NumPy
def compare_boundaries():

    print("\nBOUNDARY COMPARISON")

    years = season_df["Year"].to_numpy()
    fours = season_df["Fours"].to_numpy()
    sixes = season_df["Sixes"].to_numpy()

    # Add fours and sixes using NumPy
    total = np.add(fours, sixes)

    for i in range(len(years)):
        print(years[i], ":", total[i], "boundaries")

    # Find the highest boundary total
    highest_index = np.argmax(total)

    print("\nHighest Boundaries:", total[highest_index])
    print("Season:", years[highest_index])


# -------------------- MAIN MENU --------------------

while True:

    print("\n================================")
    print("       IPL MATCH ANALYSIS")
    print("================================")

    print("1. View Season Summary")
    print("2. View Playoff Results")
    print("3. View Top Five Players")
    print("4. Compare Boundaries")
    print("5. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        show_season_summary()

    elif choice == "2":
        show_playoff_results()

    elif choice == "3":
        show_top_players()

    elif choice == "4":
        compare_boundaries()

    elif choice == "5":
        print("\nThank you for using IPL Match Analysis!")
        break

    else:
        print("\nInvalid choice. Please try again.")



```
